```python
def tokenize(sentence):
    nlp = spacy.load("fr_core_news_sm") # Load French model 
    doc = nlp(sentence)
    tokenized_sentence = [token.text.lower() for token in doc if token.is_alpha]
    return tokenized_sentence
```
