```python
import pandas as pd
import os
import pickle
import import_ipynb
```

```python
from text_to_linguistic_features import syllabify_sentences, tokenize
```

```python
# Read each line as a sentence
tokenized_sentences = []
transcribed_sentence = []

with open("french_sentences.txt", "r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        sentence = line.strip()
        if sentence:
            # Tokenize
            tokenized_sentence = tokenize(sentence)
            tokenized_sentences.append(tokenized_sentence)
            
            # Syllabify
            transcribed_sentence = syllabify_sentences(tokenized_sentence, language="French")
            transcribed_sentences.append(transcribed_sentence)

        # Optional: limit for testing
            if i >= 10000:
             break

# Save to .pkl
with open("data/tokenized_sentences.pkl", "wb") as f:
    pickle.dump(tokenized_sentences, f)

with open("data/transcribed_sentences.pkl", "wb") as f:
    pickle.dump(transcribed_sentences, f)

print("✅ Saved tokenized and transcribed sentences for this language to 'data/'")
```
