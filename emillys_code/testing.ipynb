```python
# Sanity Check for ID computation

import random
random.seed(42)  # for reproducibility

states = ['A', 'B', 'C']
toy_sequence = [random.choice(states) for _ in range(5)]
print(f"Toy sequence: {toy_sequence}")

model = MarkovModel(n=2)
model.build(toy_sequence, input_type = "sentences")

# Compute conditional entropy
info_density = model.compute_conditional_entropy()
print(f"Conditional entropy H(Y|X) for the toy model: {information_density} bits")
```
