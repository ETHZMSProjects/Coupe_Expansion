```python
import pandas as pd
import numpy as np
import os
from cond_entropy_yoon_mi_oh_reverse_engineered import compute_cond_entropy, get_info_rate


languages = ['FRA', 'DEU','JPN', 'CMN', 'VIE', 'YUE', 'ENG']

# Load configuration CSV
config_df = pd.read_json("C:/Users/emill/Documents/GitHub/Coupe_Expansion/emillys_code/language_config.json")
config_df.set_index("Language", inplace=True)

file_path = "C:/Users/emill/Documents/GitHub/Coupe_Expansion/AutomaticSylDetect.csv"
speech_df = pd.read_csv(file_path, sep="\t")  # Assuming the file is tab-separated


# Load results CSV
csv_path = "C:/Users/emill/Documents/GitHub/Coupe_Expansion/emillys_code/produced_data/coupe_replication.csv"

# Create empty DataFrame if file doesn't exist
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path, encoding="utf-8")
else:
    df = pd.DataFrame(columns=["Language", "ID_unigram_repl", "ID_bigram_repl", "IR_bigram_repl", "std_IR_bigram_repl"])


for language in languages:

    # Extract language-specific configuration
    lang_cfg = config_df.loc[language]
    input_path = lang_cfg["Coupe Data"]
    print("\nlanguage:", lang_cfg.name)

    # Compute entropy values
    ID_unigram, ID_bigram = compute_cond_entropy(input_path, lang_cfg)
    
    # Compute the information rate (bits per second) per speaker
    unigram_info_rates = get_info_rate(ID_unigram, speech_df, language)
    bigram_info_rates = get_info_rate(ID_bigram, speech_df, language)

    # Compute mean information rate for the given language
    mean_IR_unigram = round(np.mean(unigram_info_rates), 3)
    std_IR_unigram = round(np.std(unigram_info_rates), 3)
    mean_IR_bigram = round(np.mean(bigram_info_rates), 3)
    std_IR_bigram = round(np.std(bigram_info_rates), 3)

    print(f"{language} - ID_unigram: {ID_unigram:.3f}, ID_bigram: {ID_bigram:.3f}, IR_unigram (mean): {mean_IR_unigram}, std_IR_unigram: {std_IR_unigram}, IR_bigram (mean): {mean_IR_bigram}, std_IR_bigram: {std_IR_bigram}")
    
    # Check if language is already in the table
    if language in df['Language'].values:
        df.loc[df['Language'] == language, 'ID_unigram_repl'] = round(ID_unigram, 3)
        df.loc[df['Language'] == language, 'ID_bigram_repl'] = round(ID_bigram, 3)

        df.loc[df['Language'] == language, 'IR_unigram_repl'] = mean_IR_unigram
        df.loc[df['Language'] == language, 'std_IR_unigram_repl'] = std_IR_unigram

        df.loc[df['Language'] == language, 'IR_bigram_repl'] = mean_IR_bigram
        df.loc[df['Language'] == language, 'std_IR_bigram_repl'] = std_IR_bigram
    else:
        df = pd.concat([
            df,
            pd.DataFrame([{
                'Language': language,
                'ID_unigram_repl': round(ID_unigram, 3),
                'ID_bigram_repl': round(ID_bigram, 3),

                'IR_unigram_repl': mean_IR_unigram, 
                'std_IR_unigram_repl': std_IR_unigram,
                
                'IR_bigram_repl': mean_IR_bigram, 
                'std_IR_bigram_repl': std_IR_bigram
            }])
        ], ignore_index=True)


# Column ordering
df = df[["Language", "ID_unigram_repl", "ID_bigram_repl", 
         "IR_unigram_repl", "std_IR_unigram_repl", 
         "IR_bigram_repl", "std_IR_bigram_repl"]]


# Save the csv file with the updated data
df.to_csv(csv_path, index=False, encoding="utf-8")
```
