```python
import pandas as pd
import numpy as np
import os
from helpers import get_info_rate
from cond_entropy_yoon_mi_oh_reverse_engineered import compute_cond_entropy


languages = ['JPN', 'CMN', 'VIE', 'YUE', 'ENG']

language_paths_words = {
    "FRA": "Z:/data/FRA/Lexique383.tsv",
    "JPN": "Z:/data/JPN/jpn.txt",
    "CMN": "Z:/data/CMN/cmn.txt",
    "VIE": "Z:/data/VIE/vie.txt",
    "YUE": "Z:/data/YUE/yue.txt", 
    "ENG": "Z:/data/ENG/eng.txt",
    "DEU": "Z:/data/DEU/WebCelex_German.xlsx",
}

# Load your results CSV
csv_path = "C:/Users/emill/Documents/GitHub/Coupe_Expansion/emillys_code/produced_data/coupe_replication.csv"

# Create empty DataFrame if file doesn't exist
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path, encoding="utf-8")
else:
    df = pd.DataFrame(columns=["language", "ID_unigram_repl", "ID_bigram_repl", "IR_bigram_repl"])


for language in languages:

    input_path = language_paths_words[language]

    # Compute entropy valuues
    ID_unigram, ID_bigram = compute_cond_entropy(input_path)

    # Compute the information rate (bits per second) per speaker
    info_rates = get_info_rate(ID_bigram, language)

    # Compute mean information rate for the given language
    mean_IR = np.mean(info_rates)

    print(f"{language} - ID_unigram: {ID_unigram:.3f}, ID_bigram: {ID_bigram:.3f}, IR_bigram (mean): {mean_IR:.3f}")
    
    # Check if language already in the table
    if language in df['language'].values:
        df.loc[df['language'] == language, 'ID_unigram_repl'] = round(ID_unigram, 6)
        df.loc[df['language'] == language, 'ID_bigram_repl'] = round(ID_bigram, 6)
        df.loc[df['language'] == language, 'IR_bigram_repl'] = round(mean_IR, 6)
    else:
        df = pd.concat([
            df,
            pd.DataFrame([{
                'language': language,
                'ID_unigram_repl': round(ID_unigram, 6),
                'ID_bigram_repl': round(ID_bigram, 6),
                'IR_bigram_repl': round(mean_IR, 6)
            }])
        ], ignore_index=True)

# Save the csv file with the updated data
df.to_csv(csv_path, index=False, encoding="utf-8")
```
